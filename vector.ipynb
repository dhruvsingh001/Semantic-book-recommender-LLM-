```python
books["tagged_description"].to_csv("tagged_description.txt",
                                   lineterminator = "\n",
                                   index = False,
                                   header = False)
```

```python
raw_documents = TextLoader("tagged_description.txt",encoding="utf_8").load()
text_splitter = CharacterTextSplitter(chunk_size=5, chunk_overlap=0, separator="\n")
documents = text_splitter.split_documents(raw_documents)
```

```python
documents[0]
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
embedding_model = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001"
)
```

```python
import time
from langchain_chroma import Chroma

batch_size = 50
db_books = None

for i in range(0, len(documents), batch_size):

    batch = documents[i:i + batch_size]

    if db_books is None:
        db_books = Chroma.from_documents(
            batch,
            embedding=embedding_model
        )
    else:
        db_books.add_documents(batch)

    print(f"Processed {min(i + batch_size, len(documents))} / {len(documents)} books")

    time.sleep(5)
```

```python
stored_count = db_books._collection.count()

print("Already stored:", stored_count)
print("Remaining:", len(documents) - stored_count)
```

```python
query = "A book to teach children about nature"

docs = db_books.similarity_search(query, k=10)

docs
```

```python
books[books["isbn13"] == int(docs[0].page_content.split()[0].strip('"'))]
```

```python
def retrieve_semantic_recommendations(
        query: str,
        top_k: int = 10,
) -> pd.DataFrame:
    recs = db_books.similarity_search(query, k = 20)

    books_list = []

    for i in range(0, len(recs)):
        books_list += [int(recs[i].page_content.strip('"').split()[0])]

    return books[books["isbn13"].isin(books_list)]
```

```python

retrieve_semantic_recommendations("A book to teach children about nature")
```

```python
import time
from tqdm import tqdm

# Exclude the first 700 books
new_documents = documents[900:]

# Process remaining documents in batches
batch_size = 50

for i in tqdm(range(0, len(new_documents), batch_size)):
    batch = new_documents[i:i + batch_size]

    db_books.add_documents(batch)

    time.sleep(5)

print("All new embeddings added!")
print("Total records in ChromaDB:", db_books._collection.count())
```

```python
print("Total documents:", db_books._collection.count())
```

```python
from langchain_chroma import Chroma

persistent_db = Chroma(
    collection_name="books",
    embedding_function=embedding_model,
    persist_directory="./chroma_db"
)
```

```python
data = db_books.get(
    include=["embeddings", "documents", "metadatas"]
)

batch_size = 200

for i in range(0, len(data["ids"]), batch_size):
    end = i + batch_size

    persistent_db._collection.add(
        ids=data["ids"][i:end],
        embeddings=data["embeddings"][i:end].tolist(),
        documents=data["documents"][i:end],
        metadatas=data["metadatas"][i:end]
    )

print("Existing embeddings saved successfully!")
```
